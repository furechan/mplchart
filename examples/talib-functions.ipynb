# Plotting TA-Lib indicators

TA-Lib indicators are compatible with both pandas nad polars backends.

```python
from mplchart.chart import Chart
from mplchart.samples import sample_prices
from mplchart.primitives import Candlesticks, OHLC, Volume, LinePlot, Pane

from talib.abstract import Function

prices = sample_prices(backend="polars")
```

## Candlesticks with SMA, RSI and MACD

```python
title = "Candlesticks with SMA, RSI, and MACD"

Chart(prices, title=title, max_bars=250).plot(
    Candlesticks(),
    Function("SMA", 50),
    Function("SMA", 200),
    Pane("above"),
    Function("RSI"),
    Pane("below"),
    Function("MACD")
).show()
```

## Bollinger Bands

```python
title = "Bollinger Bands"

Chart(prices, title=title, max_bars=150).plot(
    Candlesticks(),
    Function("BBANDS", 20),
).show()
```

## Stochastic Oscillator

```python
title = "Stochastic Oscillator"

Chart(prices, title=title, max_bars=250).plot(
    Candlesticks(),
    Pane("below"),
    Function("STOCH")
).show()
```

## Parabolic Stop and Reverse

```python
title = "Parabolic Stop and Reverse"

Chart(prices, title=title, max_bars=150).plot(
    Candlesticks(),
    Function("SAR") @ LinePlot(style="marker"),
).show()
```

## Hilbert Transform

```python
title = "Hilbert Transform"

Chart(prices, title=title, max_bars=150).plot(
    Candlesticks(),
    Pane("below"),
    Function("HT_SINE"),
    Pane("below"),
    Function("HT_DCPHASE"),
    Pane("below"),
    Function("HT_DCPERIOD")
).show()
```
