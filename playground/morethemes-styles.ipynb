# morethemes with mplchart

With [morethemes](https://pypi.org/project/morethemes/) installed, `Chart(style="mt:economist")` renders any named theme — the `mt:` prefix converts it into an mplchart `Styler` on the fly (`load_mt_style(name)` is the underlying function, returning the `Styler` to inspect or layer overrides on). The theme is the complete look — its rcParams, no mplchart opinions — scoped to the chart instead of set globally via `morethemes.set_theme`.

Note: many themes reference fonts that may not be installed (Volkhov, VT323, ...) — matplotlib falls back silently here.

```python
import logging

import morethemes as mt

from mplchart.chart import Chart
from mplchart.indicators import SMA
from mplchart.primitives import Candlesticks, Volume
from mplchart.samples import sample_prices

logging.getLogger("matplotlib.font_manager").setLevel(logging.ERROR)

prices = sample_prices()
list(mt.ALL_THEMES)
```

```python
def show_theme(prices, name):
    """Render a chart under a morethemes theme via the mt: style prefix."""
    chart = Chart(prices, figsize=(12, 4), max_bars=250, title=f"mt:{name}", style=f"mt:{name}")
    chart.plot(Candlesticks(), Volume(), SMA(50))
    chart.show()
```

## All themes

```python
for name in mt.ALL_THEMES:
    show_theme(prices, name)
```
