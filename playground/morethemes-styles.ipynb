# morethemes with mplchart

With [morethemes](https://pypi.org/project/morethemes/) installed, `load_mt_theme` converts a named theme into an mplchart `Styler` — `Chart(style="mt:economist")` is the string form of the same conversion. The theme is the complete look — its rcParams, no mplchart opinions — scoped to the chart instead of set globally via `morethemes.set_theme`.

Note: many themes reference fonts that may not be installed (Volkhov, VT323, ...) — matplotlib falls back silently here.

```python
import logging

import morethemes as mt

from mplchart.chart import Chart
from mplchart.indicators import SMA
from mplchart.primitives import Candlesticks, Volume
from mplchart.samples import sample_prices
from mplchart.styles.morethemes import load_mt_theme

logging.getLogger("matplotlib.font_manager").setLevel(logging.ERROR)

prices = sample_prices()
list(mt.ALL_THEMES)
```

```python
def show_theme(prices, name):
    """Render a chart under a morethemes theme converted via load_mt_theme."""
    style = load_mt_theme(name)
    chart = Chart(prices, figsize=(12, 4), max_bars=250, title=f"load_mt_theme({name!r})", style=style)
    chart.plot(Candlesticks(), Volume(), SMA(50))
    chart.show()
```

## All themes

```python
for name in mt.ALL_THEMES:
    show_theme(prices, name)
```
