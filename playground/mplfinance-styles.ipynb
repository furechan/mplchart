# mplfinance styles with mplchart

With [mplfinance](https://pypi.org/project/mplfinance/) installed, `Chart(style="mpf:yahoo")` renders any named mpf style — the `mpf:` prefix converts it into an mplchart `Styler` on the fly (`load_mpf_style(name)` is the underlying function, returning the `Styler` to inspect or layer overrides on). The rc half follows mplfinance's own application semantics (`base_mpl_style` as the stylesheet, then `rc` and the face/grid keys), `marketcolors` map onto the candle/ohlc/volume settings, and `mavcolors` become a shared `overlay.color` cycle — so moving averages draw the mpf palette in order.

Note: `y_on_right` is not mapped — axis placement is chart layout, not a look.

```python
import mplfinance as mpf

from mplchart.chart import Chart
from mplchart.indicators import SMA
from mplchart.primitives import Candlesticks, Volume
from mplchart.samples import sample_prices

prices = sample_prices()
mpf.available_styles()
```

```python
def show_style(prices, name):
    """Render a chart under an mplfinance style via the mpf: style prefix."""
    chart = Chart(prices, figsize=(12, 4), max_bars=250, title=f"mpf:{name}", style=f"mpf:{name}")
    chart.plot(Candlesticks(), Volume(), SMA(20), SMA(50))
    chart.show()
```

## All styles

```python
for name in mpf.available_styles():
    show_style(prices, name)
```
