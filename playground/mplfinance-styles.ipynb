# mplfinance styles with mplchart

With [mplfinance](https://pypi.org/project/mplfinance/) installed, `load_mpf_style` converts a named mpf style (or a `make_mpf_style` dict) into an mplchart `Styler` — `Chart(style="mpf:yahoo")` is the string form of the same conversion. The rc half follows mplfinance's own application semantics (`base_mpl_style` as the stylesheet, then `rc` and the face/grid keys), `marketcolors` map onto the candle/ohlc/volume settings, and `mavcolors` become a shared `overlay.color` cycle — so moving averages draw the mpf palette in order.

Note: `y_on_right` is not mapped — axis placement is chart layout, not a look.

```python
import mplfinance as mpf

from mplchart.chart import Chart
from mplchart.indicators import SMA
from mplchart.primitives import Candlesticks, Volume
from mplchart.samples import sample_prices
from mplchart.styles.mplfinance import load_mpf_style

prices = sample_prices()
mpf.available_styles()
```

```python
def show_style(prices, name):
    """Render a chart under an mplfinance style converted via load_mpf_style."""
    style = load_mpf_style(name)
    chart = Chart(prices, figsize=(12, 4), max_bars=250, title=f"load_mpf_style({name!r})", style=style)
    chart.plot(Candlesticks(), Volume(), SMA(20), SMA(50))
    chart.show()
```

## All styles

```python
for name in mpf.available_styles():
    show_style(prices, name)
```
