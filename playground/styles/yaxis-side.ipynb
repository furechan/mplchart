# yaxis side

The `Chart(yaxis_right=...)` option sets the side of the pane y-axis labels. The default `None` consults the `yaxis.right` style setting, else falls back to matplotlib's left convention — the shipped styles declare `yaxis.right: True` (the finance convention), so the default look renders right while styles without an opinion (a plain matplotlib stylesheet) stay left. `yaxis_right=False` moves the labels left; twin overlays (Volume on the main pane) take the opposite side.

```python
from mplchart.chart import Chart
from mplchart.indicators import SMA, RSI
from mplchart.primitives import Candlesticks, Volume
from mplchart.samples import sample_prices

prices = sample_prices()


def show_chart(prices, yaxis_right=None):
    """Render the reference chart with the given y-axis side."""
    chart = Chart(prices, figsize=(12, 6), max_bars=250, title=f"yaxis_right={yaxis_right!r}", yaxis_right=yaxis_right)
    chart.plot(Candlesticks(), Volume(), SMA(50), RSI(14))
    chart.show()
```

## Default — labels on the right

`yaxis_right=None` consults the style: the default `mplchart` style declares `yaxis.right: True`. The Volume twin overlay keeps its scale on the left.

```python
show_chart(prices)
```

## Left — the matplotlib convention

`yaxis_right=False` moves every pane's labels to the left; the Volume twin overlay flips to the right. A style can say the same via the `yaxis.right` setting — the explicit param wins over the setting.

```python
show_chart(prices, yaxis_right=False)
```
