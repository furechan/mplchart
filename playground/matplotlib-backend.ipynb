```python
%matplotlib inline

import yfinance as yf

from mplchart.chart import Chart
from mplchart.primitives import Candlesticks, Pane, Volume
from mplchart.indicators import ROC, SMA, EMA, RSI, MACD
from mplchart.utils import normalize_prices
```

```python

ticker = 'AAPL'
prices = normalize_prices(yf.Ticker(ticker).history('5y'))
prices
```

```python

Chart(prices, title=ticker, max_bars=250).plot(
    Candlesticks(), SMA(50), SMA(200), Volume(),
    Pane("above"), RSI(),
    Pane("below"), MACD(),
).show()
```
