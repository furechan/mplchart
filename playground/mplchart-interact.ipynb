# Using mplchart interactively


```python
import yfinance as yf

from functools import lru_cache
from mplchart.utils import normalize_prices
from mplchart.notebook import chart_widget
from mplchart.primitives import Volume, Candlesticks, Pane
from mplchart.indicators import SMA, RSI, PPO


@lru_cache(maxsize=32)
def get_prices(ticker: str, period: str = "5y"):
    return normalize_prices(yf.Ticker(ticker).history(period=period))
```

```python
indicators = [Candlesticks(), SMA(50), SMA(200), Volume(), Pane("above"), RSI(), Pane("below"), PPO()]

chart_widget(get_prices, ticker="AAPL", max_bars=250, indicators=indicators)
```
