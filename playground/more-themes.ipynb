# morethemes with mplchart

With [morethemes](https://pypi.org/project/morethemes/) installed, its themes resolve as mplchart styles **by name** — resolution order is shipped styles → matplotlib stylesheets → installed theme providers. The theme is the complete look (no mplchart opinions ride along). Equivalent explicit form: `style={"rc": morethemes.get_rcparams(name)}`.

Note: many themes reference fonts that may not be installed (Volkhov, VT323, ...) — matplotlib falls back silently here.

```python
import logging

import morethemes as mt

from mplchart.chart import Chart
from mplchart.indicators import SMA
from mplchart.primitives import Candlesticks, Volume
from mplchart.samples import sample_prices

logging.getLogger("matplotlib.font_manager").setLevel(logging.ERROR)

prices = sample_prices()
list(mt.ALL_THEMES)
```

```python
def show_theme(prices, name):
    """Render a chart under a morethemes theme, resolved by name."""
    chart = Chart(prices, figsize=(12, 4), max_bars=250, title=f"style={name!r} (morethemes)", style=name)
    chart.plot(Candlesticks(), Volume(), SMA(50))
    chart.show()
```

## All themes

```python
for name in mt.ALL_THEMES:
    show_theme(prices, name)
```
